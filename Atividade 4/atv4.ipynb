Exercício 3 — Número de soluções

a) $\begin{cases}x+y=3\\2x+2y=6\end{cases}$

$$\left[\begin{array}{rr|r}1&1&3\\2&2&6\end{array}\right]\xrightarrow{L_2\leftarrow L_2-2L_1}\left[\begin{array}{rr|r}1&1&3\\0&0&0\end{array}\right]$$
A segunda linha é nula ($0=0$): há apenas 1 equação independente para 2 incógnitas. 
Infinitas soluções: $x=3-t,\ y=t,\ t\in\mathbb{R}$. ($\text{posto}(A)=\text{posto}([A|b])=1<2$)

b) $\begin{cases}x+y=3\\2x+2y=7\end{cases}$

$$\left[\begin{array}{rr|r}1&1&3\\2&2&7\end{array}\right]\xrightarrow{L_2\leftarrow L_2-2L_1}\left[\begin{array}{rr|r}1&1&3\\0&0&1\end{array}\right]$$
A segunda linha representa $0=1$, impossível. Nenhuma solução (sistema inconsistente). ($\text{posto}(A)=1<\text{posto}([A|b])=2$)

c) $\begin{cases}x+y=3\\2x-y=0\end{cases}$

$$\left[\begin{array}{rr|r}1&1&3\\2&-1&0\end{array}\right]\xrightarrow{L_2\leftarrow L_2-2L_1}\left[\begin{array}{rr|r}1&1&3\\0&-3&-6\end{array}\right]$$
$-3y=-6\Rightarrow y=2$; $x=3-2=1$. Solução única: $(x,y)=(1,2)$. ($\text{posto}(A)=\text{posto}([A|b])=2=n$)

Verificação computacional pelo critério dos postos (Rouché–Capelli):

```python
def classificar(A, b):
    # Classifica o sistema comparando posto(A), posto([A|b]) e o número de incógnitas n
    A = np.array(A, dtype=float); b = np.array(b, dtype=float)
    n = A.shape[1]
    rA  = np.linalg.matrix_rank(A)
    rAb = np.linalg.matrix_rank(np.column_stack([A, b]))
    if rA < rAb:
        tipo = "nenhuma solução (inconsistente)"
    elif rA == n:
        tipo = "solução única"
    else:
        tipo = "infinitas soluções"
    return rA, rAb, tipo

sistemas = {
    "a)": ([[1, 1], [2,  2]], [3, 6]),
    "b)": ([[1, 1], [2,  2]], [3, 7]),
    "c)": ([[1, 1], [2, -1]], [3, 0]),
}
for nome, (A, b) in sistemas.items():
    rA, rAb, tipo = classificar(A, b)
    print(f"{nome} posto(A)={rA}, posto([A|b])={rAb}  ->  {tipo}")
```

Interpretação: os resultados computacionais confirmam a análise manual. Geometricamente: em (a) as duas retas são coincidentes, em (b) são paralelas distintas e em (c) se cruzam em um único ponto.

Exercício 4 — Eliminação de Gauss sem pivoteamento

```python
def gauss_sem_pivoteamento(A, b, verbose=False):
    '''Resolve Ax = b por eliminação de Gauss SEM pivoteamento.

    Parâmetros
    ----------
    A : array (n, n)  matriz dos coeficientes
    b : array (n,)    vetor independente
    verbose : se True, imprime a matriz aumentada a cada etapa

    Retorna
    -------
    x : array (n,)    solução calculada
    '''
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    n = len(b)

    # 1) Forma a matriz aumentada [A | b]
    Ab = np.column_stack([A, b])
    if verbose:
        print("Matriz aumentada inicial:\n", Ab, "\n")

    # 2) Eliminação: zera os elementos abaixo de cada pivô
    for k in range(n - 1):
        if Ab[k, k] == 0:
            raise ZeroDivisionError(f"Pivô nulo na posição ({k},{k}): "
                                    "é necessário pivoteamento.")
        for i in range(k + 1, n):
            m = Ab[i, k] / Ab[k, k]           # multiplicador m_ik
            Ab[i, k:] = Ab[i, k:] - m * Ab[k, k:]   # L_i <- L_i - m * L_k
            Ab[i, k] = 0.0                    # evita resíduo de arredondamento
        if verbose:
            print(f"Após eliminar a coluna {k+1}:\n", Ab, "\n")

    if Ab[n-1, n-1] == 0:
        raise ZeroDivisionError("Pivô nulo no último elemento: sistema singular.")

    # 3) Substituição regressiva
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        soma = np.dot(Ab[i, i+1:n], x[i+1:n])
        x[i] = (Ab[i, n] - soma) / Ab[i, i]
    return x
```

Teste com o sistema do Exercício 2 e comparação com numpy.linalg.solve:

```python
A2 = np.array([[ 2,  1, -1],
               [-3, -1,  2],
               [-2,  1,  2]], dtype=float)
b2 = np.array([8, -11, -3], dtype=float)

x_gauss = gauss_sem_pivoteamento(A2, b2, verbose=True)
x_numpy = np.linalg.solve(A2, b2)

print("Solução (Gauss implementado):", x_gauss)
print("Solução (numpy.linalg.solve):", x_numpy)
print("Diferença máxima |x_gauss - x_numpy| =", np.max(np.abs(x_gauss - x_numpy)))
print("Resíduo ||A x - b||∞ =", np.max(np.abs(A2 @ x_gauss - b2)))
```

Interpretação: as matrizes aumentadas impressas coincidem com as etapas feitas à mão no Exercício 2 (após a etapa 1: linhas $[0,\ 0.5,\ 0.5\,|\,1]$ e $[0,\ 2,\ 1\,|\,5]$; após a etapa 2: $[0,\ 0,\ -1\,|\,1]$). A solução $(2,3,-1)$ concorda com a manual e com o numpy.linalg.solve (que usa LU com pivoteamento parcial) a menos de erro de arredondamento da ordem de $10^{-15}$. Neste sistema os pivôs ($2$, $0.5$, $-1$) nunca se anulam, por isso o método sem pivoteamento funciona; em geral, um pivô nulo ou muito pequeno exigiria pivoteamento.

Exercício 5 — Modelo de balanço térmico

$$\begin{bmatrix}4&-1&0\\-1&4&-1\\0&-1&3\end{bmatrix}\begin{bmatrix}T_1\\T_2\\T_3\end{bmatrix}=\begin{bmatrix}15\\10\\10\end{bmatrix}$$

a) Resolução manual

Matriz aumentada:
$$\left[\begin{array}{rrr|r}4&-1&0&15\\-1&4&-1&10\\0&-1&3&10\end{array}\right]$$

Etapa 1 — pivô $4$; $m_{21}=-1/4$, $m_{31}=0$: $L_2\leftarrow L_2+\tfrac14L_1$
$$\left[\begin{array}{rrr|r}4&-1&0&15\\0&\tfrac{15}{4}&-1&\tfrac{55}{4}\\0&-1&3&10\end{array}\right]$$

Etapa 2 — pivô $\tfrac{15}{4}$; $m_{32}=-1/(15/4)=-\tfrac{4}{15}$: $L_3\leftarrow L_3+\tfrac{4}{15}L_2$
$$\left[\begin{array}{rrr|r}4&-1&0&15\\0&\tfrac{15}{4}&-1&\tfrac{55}{4}\\0&0&\tfrac{41}{15}&\tfrac{41}{3}\end{array}\right]$$
(pois $3-\tfrac{4}{15}=\tfrac{41}{15}$ e $10+\tfrac{4}{15}\cdot\tfrac{55}{4}=10+\tfrac{11}{3}=\tfrac{41}{3}$)

Substituição regressiva:
- $\tfrac{41}{15}T_3=\tfrac{41}{3}\Rightarrow T_3=5$
- $\tfrac{15}{4}T_2-5=\tfrac{55}{4}\Rightarrow \tfrac{15}{4}T_2=\tfrac{75}{4}\Rightarrow T_2=5$
- $4T_1-5=15\Rightarrow T_1=5$

$$\boxed{T_1=T_2=T_3=5}$$

Conferência: $4(5)-5=15$; $-5+20-5=10$; $-5+15=10$


b) Implementação em Python

```python
A5 = np.array([[ 4, -1,  0],
               [-1,  4, -1],
               [ 0, -1,  3]], dtype=float)
b5 = np.array([15, 10, 10], dtype=float)

T_gauss = gauss_sem_pivoteamento(A5, b5, verbose=True)   # nossa implementação (Ex. 4)
T_numpy = np.linalg.solve(A5, b5)                        # referência

print("T (Gauss implementado):", T_gauss)
print("T (numpy.linalg.solve):", T_numpy)
print("Diferença máxima:", np.max(np.abs(T_gauss - T_numpy)))
print("Resíduo ||A T - b||∞ =", np.max(np.abs(A5 @ T_gauss - b5)))
```

Interpretação: a solução numérica $T_1=T_2=T_3=5$ coincide com a manual. A matriz é tridiagonal e diagonalmente dominante (com dominância estrita na 1ª e na 3ª linha), o que garante pivôs não nulos e estabilidade da eliminação sem pivoteamento. Fisicamente, as três temperaturas se igualam em equilíbrio porque as fontes (lado direito) são compatíveis com os acoplamentos da rede: $15 = 4\cdot5-5$, $10=-5+20-5$, $10=-5+15$.
